# Packages import

```python
import os
import csv
from datetime import datetime
from zoneinfo import ZoneInfo
from ics import Calendar, Event
```

## Calendar export

```python
file_name = input("Enter the name of the CSV file (without .csv extension): ")
```

```python
calendar = Calendar()
# calendar["dstart"] = "TZID=Europe/Warsaw:20260223T000000"
# calendar["dtend"] = "TZID=Europe/Warsaw:20260930T235999"
# calendar["summary"] = f"Schedule for {file_name}"
```

```python
with open(f"./schedules/{file_name}.csv", encoding="utf-8") as csvfile:
    reader = csv.DictReader(csvfile)

    local_tz = ZoneInfo("Europe/Warsaw")
    for row in reader:
        event = Event()
        event.name = row["Przedmiot"]
        start_initial = datetime.strptime(row["Termin"] + " " + row["Start time"], "%Y-%m-%d %H:%M")
        end_initial = datetime.strptime(row["Termin"] + " " + row["End time"], "%Y-%m-%d %H:%M")

        event.begin = start_initial.replace(tzinfo=local_tz)
        event.end = end_initial.replace(tzinfo=local_tz)

        event.location = row["Sala"]
        event.description = f"Type: {row['Typ']}, Teacher: {row['Nauczyciel']}, Duration: {row['Duration']}"
        calendar.events.add(event)
    
    if not os.path.exists("./calendars"):
        os.makedirs("./calendars")

    with open(f"./calendars/{file_name}.ics", "w", encoding="utf-8") as icsfile:
        icsfile.writelines(calendar.serialize())
```
